```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int , b: int):
    """ it will return the sum of two numbers 

        Args:
            a : number one
            b : number two
    """
    return a + b

@tool
def multiply_numbers(a: int , b: int):
    """ it will return the product of two numbers 

        Args:
            a : number one
            b : number two
    """
    return a + b
```

```python
add_numbers.invoke({"a": 5, "b": 3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="you are a math teacher always use tool for calculation."
)
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"what is 2+3 ?"}]})
```
